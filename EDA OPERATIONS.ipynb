```python
import pandas as pd

df = pd.read_csv("C:\\Users\\satya\\Downloads\\archive (1)\\AB_NYC_2019.csv")

print("Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

df.head(3)
```

```python
df_clean = df.drop(columns=["name", "host_name", "last_review"])

print("New shape:", df_clean.shape)
df_clean.head(3)
```

```python
df_clean["reviews_per_month"] = df_clean["reviews_per_month"].fillna(0)

print("Missing reviews_per_month:", df_clean["reviews_per_month"].isna().sum())
df_clean.head(3)
```

```python
p99 = df_clean["price"].quantile(0.99)
print("99th percentile price:", p99)
```

```python
df_clean = df_clean[(df_clean["price"] > 0) & (df_clean["price"] <= 799)]

print("New shape after removing outliers:", df_clean.shape)
df_clean["price"].describe()
```

```python
df_model = df_clean.drop(columns=["id", "host_id"])

print("Columns used for model:")
print(df_model.columns.tolist())

print("\nShape:", df_model.shape)
```

```python
X = df_model.drop(columns=["price"])
y = df_model["price"]

print("Feature shape:", X.shape)
print("Target shape:", y.shape)

X.head(3)
```

```python
print("Categorical columns:")
print(X.select_dtypes(include="object").columns.tolist())

print("\nNumerical columns:")
print(X.select_dtypes(exclude="object").columns.tolist())
```

```python
X_encoded = pd.get_dummies(X, drop_first=True)

print("New shape after encoding:", X_encoded.shape)
X_encoded.head(3)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y, 
    test_size=0.2, 
    random_state=42
)

print("Training set shape:", X_train.shape)
print("Testing set shape:", X_test.shape)
```

```python
from sklearn.linear_model import LinearRegression

# Create model
lr_model = LinearRegression()

# Train model
lr_model.fit(X_train, y_train)

print("Model training completed.")
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Predict on test set
y_pred = lr_model.predict(X_test)

# Calculate metrics
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("MAE:", round(mae, 2))
print("RMSE:", round(rmse, 2))
print("R2 Score:", round(r2, 4))
```

```python
from sklearn.ensemble import RandomForestRegressor

rf_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train, y_train)

print("Random Forest training completed.")
```

```python
# Predict using Random Forest
y_pred_rf = rf_model.predict(X_test)

# Calculate metrics
mae_rf = mean_absolute_error(y_test, y_pred_rf)
rmse_rf = np.sqrt(mean_squared_error(y_test, y_pred_rf))
r2_rf = r2_score(y_test, y_pred_rf)

print("Random Forest Results")
print("MAE:", round(mae_rf, 2))
print("RMSE:", round(rmse_rf, 2))
print("R2 Score:", round(r2_rf, 4))
```

```python
# Predict using Random Forest on entire dataset
df_model["ml_predicted_price"] = rf_model.predict(X_encoded)

df_model[["price", "ml_predicted_price"]].head()
```

```python
# Occupancy proxy (lower availability = higher demand)
df_model["occupancy_proxy"] = 1 - (df_model["availability_365"] / 365)

# Normalize number_of_reviews
df_model["reviews_norm"] = (
    df_model["number_of_reviews"] - df_model["number_of_reviews"].min()
) / (
    df_model["number_of_reviews"].max() - df_model["number_of_reviews"].min()
)

# Create demand index (weighted)
df_model["demand_index"] = (
    0.6 * df_model["occupancy_proxy"] +
    0.4 * df_model["reviews_norm"]
)

df_model[["occupancy_proxy", "reviews_norm", "demand_index"]].head()
```

```python
# Define max dynamic adjustment (30%)
max_adjustment = 0.30

# Calculate dynamic suggested price
df_model["dynamic_suggested_price"] = (
    df_model["ml_predicted_price"] *
    (1 + df_model["demand_index"] * max_adjustment)
)

df_model[["price", "ml_predicted_price", "demand_index", "dynamic_suggested_price"]].head()
```

```python
# Calculate uplift %
df_model["uplift_percentage"] = (
    (df_model["dynamic_suggested_price"] - df_model["price"])
    / df_model["price"]
) * 100

df_model[["price", "dynamic_suggested_price", "uplift_percentage"]].head()
```

```python
# Average uplift
avg_uplift = df_model["uplift_percentage"].mean()

# Listings that can increase price
increase_count = (df_model["uplift_percentage"] > 5).sum()

# Listings that should reduce price
decrease_count = (df_model["uplift_percentage"] < -5).sum()

print("Average Uplift %:", round(avg_uplift, 2))
print("Listings with Increase Opportunity:", increase_count)
print("Listings Overpriced (Should Reduce):", decrease_count)
```

```python
# Cap uplift at -50% and +50%
df_model["uplift_capped"] = df_model["uplift_percentage"].clip(-50, 50)

# Recalculate final recommended price based on capped uplift
df_model["final_recommended_price"] = (
    df_model["price"] * (1 + df_model["uplift_capped"] / 100)
)

df_model[["price", "uplift_percentage", "uplift_capped", "final_recommended_price"]].head()
```

```python
comparison = pd.DataFrame({
    "Current_Avg_Price": [df_model["price"].mean()],
    "ML_Base_Avg_Price": [df_model["ml_predicted_price"].mean()],
    "Dynamic_Avg_Price": [df_model["dynamic_suggested_price"].mean()],
    "Final_Recommended_Avg": [df_model["final_recommended_price"].mean()]
})

comparison.round(2)
```
